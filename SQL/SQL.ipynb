```python
import sqlite3
import pandas as pd
```

```python
# create connection
conn = sqlite3.connect("company.db")
print(conn)
```

```python
# create cursor
cur = conn.cursor()
print(cur)
```

```python
# create table:
# TABLE_NAME(
#    COL1 TYPE,
#    COL2 TYPE
# )

cur.execute('''
CREATE TABLE IF NOT EXISTS employees(
    emp_id INTEGER PRIMARY KEY,
    full_name TEXT NOT NULL,
    department TEXT,
    salary REAL
)
''')
print("table created")
```

```python
# insert one row
cur.execute(
    "INSERT INTO employees (emp_id, full_name, department, salary)"
    "VALUES (?, ?, ?, ?)",
    (1, "Amara Johnson", "Finance", 96500.0)
)
```

```python
cur.execute("SELECT * FROM employees WHERE emp_id = 1").fetchall()
```

```python
# insert many rows at once
rows = [
    (2, "Daniel Osei", "Sales", 61000.0),
    (3, "Fatima Bah", "Engineering", 142000.0),
    (4, "Marcus Lee", "Engineerng", 118000.0),
]

cur.executemany(
    "INSERT INTO employees (emp_id, full_name, department, salary)"
    "VALUES (?, ?, ?, ?)",
    rows
)
```

```python
# read data back
cur.execute("SELECT * FROM employees ORDER BY emp_id").fetchall()
```

```python
conn.rollback()
df = pd.read_sql_query("SELECT emp_id, full_name, salary FROM employees ORDER BY emp_id", conn)
df
```

```python
conn.rollback()
df = pd.read_sql_query("SELECT * FROM employees ORDER BY emp_id", conn)
df
```

```python
# save changes
conn.commit()
print("saved")
```
