## Problem 7

```python
conn.row_factory = sqlite3.Row # rows become dict-like
c2 = conn.cursor()
c2.execute("""SELECT full_name, department, hired_on, region, phone
            FROM employees
            WHERE department IN (?, ?) 
            AND hired_on BETWEEN ? AND ?
            AND phone NOT LIKE ?
            AND region IS NOT NULL""", ('Finance', 'Engineering', '2018-01-01', '2022-12-31', '312%'))
for r in c2:
    print(f" {r['full_name']:<16}{r['department']:<16}{r['hired_on']:<16}"
            f"{r['region']:<16}{r['phone']:>16}")
conn.row_factory = None # back to plain tuples
```

```python
def find_staff(depts, start, end):
    conn.row_factory = sqlite3.Row # rows become dict-like
    c2 = conn.cursor()
    params = list(depts) + [start, end, '312%']
    c2.execute(f"""SELECT full_name, department, hired_on, region, phone
                FROM employees
                WHERE department IN ({", ".join("?" *len(depts))}) 
                AND hired_on BETWEEN ? AND ?
                AND phone NOT LIKE ?
                AND region IS NOT NULL""", params)

    for r in c2:
        print(f" {r['full_name']:<16}{r['department']:<16}{r['hired_on']:<16}"
                f"{r['region']:<16}{r['phone']:>16}")
    conn.row_factory = None # back to plain tuples
    
find_staff(['Finance', 'Engineering'], '2018-01-01', '2022-12-31')
```
