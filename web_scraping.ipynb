```python
import sys, requests, bs4, lxml, pandas as pd
```

```python
url = "https://en.wikipedia.org/wiki/List_of_S%26P_500_companies"

r = requests.get(url, timeout=30)

r.raise_for_status()

print(r.status_code) # 200

print(len(r.text)) # characters of HTML

print(r.text[:300]) # look at it!
```

```python
url = "https://en.wikipedia.org/wiki/List_of_S%26P_500_companies"

headers = {

"User-Agent": "HowardU-DataWrangling/1.0 (+https://howard.edu/course)",

"Accept-Encoding": "gzip, deflate",

}

r = requests.get(url, headers=headers, timeout=30)

r.raise_for_status()

print(r.status_code) # 200

print(len(r.text)) # characters of HTML

print(r.text[:300]) # look at it!
```

```python
import urllib.robotparser

rp = urllib.robotparser.RobotFileParser()

rp.set_url("https://en.wikipedia.org/robots.txt")

rp.read()

rp.can_fetch("HowardU-DataWrangling/1.0", "https://en.wikipedia.org/wiki/List_of_S%26P_500_companies")
```
