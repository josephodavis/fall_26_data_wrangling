```python
import pandas as pd
import numpy as np
from pathlib import Path

PATH = Path("math153_signups_raw.csv")
TODAY = pd.Timestamp("2026-10-28")

df = pd.read_csv(PATH)
print("shape:", df.shape)
df.head(5)
```

```python
missing = df.isna().mean().sort_values(ascending=False)
print("share of values missing, by column")
print((missing[missing > 0] * 100).round(1).to_string())
```

```python
print("credits_completed — most common values")
print(df["credits_completed"].value_counts().head(5).to_string())
print()
print("rows holding the sentinel -999:", (df["credits_completed"] == -999).sum())
print("minimum credits as loaded    :", df["credits_completed"].min())
print()
print("mean WITH the sentinel   :", round(df["credits_completed"].mean(), 1))
clean_credits = df["credits_completed"].replace(-999, np.nan)
print("mean WITHOUT the sentinel:", round(clean_credits.mean(), 1))
```
