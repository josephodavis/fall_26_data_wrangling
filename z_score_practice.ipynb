```python
import numpy as np

nums = [42, 45, 44, 46, 43, 45, 47, 44, 43, 46, 
 45, 43, 44, 46, 45, 44, 43, 47, 45, 150]
```

```python
mean = np.mean(nums)
std = np.std(nums)
print(mean)
print(std)
```

```python
# 150 z score

z = (150 - mean) / std
print(z)
# greater than 3
# (n-1)/root(n) = 19/root(20) = 4.24 --> answer quite close, but still too high
```

```python
b_nums = [18, 20, 19, 21, 20, 22, 19, 90]

q1, q3 = np.percentile(b_nums, [25, 75])

iqr = q3 - q1

low = q1 - (1.5*iqr)
high = q3 + (1.5*iqr)

print("flagged:")
for num in b_nums:
    if num < low or num > high:
        print(num)
```

```python
c_nums = [26, 27, 28, 29, 30, 30, 31, 31, 32, 33, 150]

c_mean = np.mean(c_nums)
c_std = np.std(c_nums)

c_z_scores = []

for num in c_nums:
    c_z_scores.append((num-c_mean)/c_std)

largest_z = max(c_z_scores)

print(largest_z)
# greater than 3
```

```python
c_med = np.median(c_nums)
c_mad = np.median(np.abs(c_nums - c_med))

M = 0.6745 * (c_nums - c_med) / c_mad

print(np.round(M, 2))
```
