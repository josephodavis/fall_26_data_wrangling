# Regex for Financial Data — Solutions

Worked solutions for `regex_finance_exercises.ipynb`. Each one explains *why* the pattern is shaped
the way it is, and several show the naive version failing next to it.

Run the notebook top to bottom — later cells reuse names defined earlier.

(?:...)     group, but don't capture

(?=...)     lookahead

(?!...)     negative lookahead

(?<=...)    lookbehind

(?#...)     comment

(?i)        inline flag

(?P<...>)   named group      <-- this one

```python
import re
import pandas as pd

pd.set_option("display.max_colwidth", 90)
pd.set_option("display.width", 200)

df = pd.read_csv("finance_transactions.csv", dtype=str).fillna("")
print(df.shape)
df.head(5)
```

## The data

`finance_transactions.csv` — 260 rows of  messy transaction records. Nothing here is clean;
that is the point. Every column mixes several real-world conventions.

| column | what's messy about it |
|---|---|
| `txn_id` | 5 competing ID conventions: `TXN-2024-0001`, `txn_2024_0001`, `TX 2024/0001`, `TXN20240001`, and a malformed short-year `TXN-24-0001` |
| `posted_date` | 6 formats: ISO, `MM/DD/YYYY`, `DD/MM/YYYY`, `Mar 15, 2024`, `DD.MM.YYYY`, `YYYYMMDD` |
| `description` | raw bank memo lines: merchant, city/state, card suffix, `REF#`, `TERM:`, `SEQ`, IBANs, star-prefixed processor names, random runs of whitespace |
| `amount_raw` | `$1,234.56`, `-$45.00`, `(89.99)` (parens = negative), `USD 1,203.40`, `1.234,56 EUR` (European decimal comma), `£45.60`, `-JPY 900.00` |
| `currency` | sometimes blank — you may have to recover it from `amount_raw` |
| `account` | `****1234`, `XXXX-XXXX-XXXX-1234`, spaced IBANs, `1234-5678-9012-3456`, `ACCT NO 1234567890` |
| `merchant_contact` | emails (some deliberately broken: `a@@b.com`, `user@nodot`, `name at domain dot com`) and phones in 7 formats |
| `channel` / `status` | inconsistent casing |
| `notes` | free text with invoice numbers, PO numbers, EINs, VAT ids, **SSNs that must be redacted**, percentages, timestamps, and accidental doubled words |

Run the cell below to eyeball the variety before you start.

```python
for col in ["txn_id", "posted_date", "amount_raw", "account", "merchant_contact"]:
    print(f"--- {col} ---")
    print(df[col].sample(8, random_state=7).to_list())
    print()

print("--- description ---")
for d in df.description.sample(6, random_state=3):
    print(" ", repr(d))
print("\n--- notes ---")
for n in df.notes.sample(6, random_state=3):
    print(" ", n)
```

## Section 1 — Anchors, groups and alternation

Validation and normalisation: the two jobs regex is actually good at.

### Solution 1

`fullmatch` requires the pattern to consume the whole string, so no `^`/`$` are needed — that is
exactly what you want for validation. `re.match` would happily accept `TXN-2024-0001-JUNK`.

```python
from collections import Counter

CANONICAL_ID = re.compile(r"TXN-2024-\d{4}")

bad_ids = [t for t in df.txn_id if not CANONICAL_ID.fullmatch(t)]
print("conforming:", len(df) - len(bad_ids), "| non-conforming:", len(bad_ids))
print(bad_ids[:8])

# group the offenders by shape
print(Counter(re.sub(r"\d", "N", t) for t in bad_ids))

# contrast: match() is not anchored on the right
print(bool(re.match(CANONICAL_ID, "TXN-2024-0001-JUNK")),
      bool(re.fullmatch(CANONICAL_ID, "TXN-2024-0001-JUNK")))
```

### Solution 2

One pattern, three named groups, optional single-character separators. Because `year` is `\d{4}`,
`TXN-24-0002` cannot match: after `TXN` and the `-`, the engine needs four digits and finds `24-0`.
That is a much better failure than silently parsing `24` as a year.

```python
ID_RE = re.compile(r"^(?P<prefix>TXN?)[-_ ]?(?P<year>\d{4})[-_/ ]?(?P<seq>\d{4})$", re.I)

def normalize_id(raw):
    m = ID_RE.match(raw.strip())
    if not m:
        return None
    return f"TXN-{m.group('year')}-{m.group('seq')}"

norm = df.txn_id.map(normalize_id)
print("normalized:", norm.notna().sum(), "| rejected:", norm.isna().sum())
print(pd.DataFrame({"raw": df.txn_id, "clean": norm}).head(6).to_string(index=False))
print("\nrejected sample:", df.txn_id[norm.isna()].head(4).to_list())
```

### Solution 3

Order matters inside an alternation: the engine takes the **first** branch that matches, so put the
more specific shapes first. `re.VERBOSE` (`re.X`) ignores unescaped whitespace and lets `#` start a
comment — which is the only reason a pattern this size stays readable.

```python
MONTHS = ["Jan", "Feb", "Mar", "Apr", "May", "Jun", "Jul", "Aug", "Sep", "Oct", "Nov", "Dec"]

DATE_RE = re.compile(r"""
    ^\s*(?:
        (?P<iso_y>\d{4})-(?P<iso_m>\d{2})-(?P<iso_d>\d{2})               # 2024-08-17
      | (?P<sl_a>\d{1,2})/(?P<sl_b>\d{1,2})/(?P<sl_y>\d{4})              # 09/15/2024 or 24/03/2024
      | (?P<mon>Jan|Feb|Mar|Apr|May|Jun|Jul|Aug|Sep|Oct|Nov|Dec)
        \s+(?P<ln_d>\d{1,2}),\s*(?P<ln_y>\d{4})                          # Jan 17, 2024
      | (?P<dt_d>\d{2})\.(?P<dt_m>\d{2})\.(?P<dt_y>\d{4})                # 24.03.2024
      | (?P<cp_y>\d{4})(?P<cp_m>\d{2})(?P<cp_d>\d{2})                    # 20240317
    )\s*$
""", re.X)

def parse_date(raw):
    m = DATE_RE.match(raw)
    if not m:
        return None
    g = m.groupdict()
    if g["iso_y"]:
        return int(g["iso_y"]), int(g["iso_m"]), int(g["iso_d"])
    if g["sl_y"]:
        a, b = int(g["sl_a"]), int(g["sl_b"])
        month, day = (b, a) if a > 12 else (a, b)      # >12 can only be a day
        return int(g["sl_y"]), month, day
    if g["mon"]:
        return int(g["ln_y"]), MONTHS.index(g["mon"]) + 1, int(g["ln_d"])
    if g["dt_y"]:
        return int(g["dt_y"]), int(g["dt_m"]), int(g["dt_d"])
    return int(g["cp_y"]), int(g["cp_m"]), int(g["cp_d"])

parsed = df.posted_date.map(parse_date)
print("unparsed:", parsed.isna().sum(), "of", len(df))
df["date_iso"] = parsed.map(lambda t: f"{t[0]:04d}-{t[1]:02d}-{t[2]:02d}")
print(pd.DataFrame({"raw": df.posted_date, "iso": df.date_iso}).sample(8, random_state=1).to_string(index=False))
```

### Solution 4

Three separate optional sign slots (`sign1/2/3`) is not elegant, but it beats trying to write one
clever alternation — real memo formats are not systematic, and optional groups cost nothing when
absent. Note that deciding "is this comma a decimal point?" is *not* a regex job in the matching
step; a second, tiny anchored pattern answers it cleanly afterwards.

```python
AMOUNT_RE = re.compile(r"""
    ^\s*
    (?P<open>\()?               \s*     # ( ... ) means negative
    (?P<sign1>-)?               \s*
    (?:(?P<code1>USD|EUR|GBP|JPY)\s*)?  # leading ISO code
    (?P<sign2>-)?               \s*
    (?P<sym1>[$£€])?            \s*     # leading symbol
    (?P<sign3>-)?               \s*
    (?P<body>\d[\d.,]*)         \s*     # the digits, separators still ambiguous
    (?P<sym2>[$£€])?            \s*     # trailing symbol
    (?P<code2>USD|EUR|GBP|JPY)? \s*
    (?P<close>\))?              \s*$
""", re.X)

EURO_BODY = re.compile(r"^\d{1,3}(?:\.\d{3})*,\d{2}$")
SYMBOL_TO_CODE = {"$": "USD", "£": "GBP", "€": "EUR"}

def parse_amount(raw):
    m = AMOUNT_RE.match(raw)
    if not m:
        return None
    g = m.groupdict()
    body = g["body"]
    value = (float(body.replace(".", "").replace(",", "."))
             if EURO_BODY.match(body) else float(body.replace(",", "")))
    negative = bool(g["open"] and g["close"]) or any(g[k] for k in ("sign1", "sign2", "sign3"))
    code = g["code1"] or g["code2"] or SYMBOL_TO_CODE.get(g["sym1"] or g["sym2"] or "")
    return (-value if negative else value), code

parsed_amt = df.amount_raw.map(parse_amount)
print("unparsed:", sum(v is None for v in parsed_amt))

df["amount"] = [v for v, _ in parsed_amt]
df["ccy_from_text"] = pd.Series([c for _, c in parsed_amt], index=df.index)

# a bare "1,234.56" carries no currency signal at all - fall back to the (sometimes blank) column
print("currency read from the amount string:", df.ccy_from_text.notna().sum())
df["ccy"] = df.ccy_from_text.fillna(df.currency.replace("", pd.NA))
print("currency still unknown after fallback:", df.ccy.isna().sum())
print("blank currency cells recovered from text:",
      ((df.currency == "") & df.ccy_from_text.notna()).sum())
print(pd.DataFrame({"raw": df.amount_raw, "value": df.amount, "ccy": df.ccy})
        .sample(10, random_state=5).to_string(index=False))
```

## Section 2 — Lookarounds

Zero-width assertions: matching *because of* context without consuming it.

### Solution 5

`(?:(?<=\*\*\*\*)|(?<=CARD ))` — two fixed-width lookbehinds inside a non-capturing group. Each
alternative is individually fixed width, which satisfies the engine. The match itself is just
`\d{4}`, so nothing but the digits is consumed.

```python
CARD_RE = re.compile(r"(?:(?<=\*\*\*\*)|(?<=CARD\ ))\d{4}", re.X)

card = df.description.map(lambda s: m.group(0) if (m := CARD_RE.search(s)) else None)
print("rows with a card suffix:", card.notna().sum())
print(df.loc[card.notna(), "description"].head(3).to_list())
print(card.dropna().head(6).to_list())

# why the "obvious" version fails:
try:
    re.compile(r"(?<=\*\*\*\*|CARD )\d{4}")
except re.error as e:
    print("\nvariable-width lookbehind ->", e)
```

### Solution 6

When the left-hand context varies in length, stop fighting the lookbehind rule: consume the marker
in a group and read `group("code")`. It is also faster — a lookbehind re-tests at every position.

```python
REF_RE = re.compile(r"(?P<label>REF#|REF\s|CASE#|CONF\s|SEQ\s|ID:)(?P<code>[A-Z0-9]+)")

found = df.description.str.extract(REF_RE)
print("rows with a reference code:", found.code.notna().sum())
print(found.label.str.strip().value_counts().to_string())
print(found.dropna().head(6).to_string(index=False))
```

### Solution 7

Stacked assertions are the readable way to express "must contain X and must not contain Y" in one
pattern. Note `\b` around the words: without it, `REFUNDED` and `IRREVERSAL` would both count.

```python
REFUND_ONLY = re.compile(r"^(?=.*\bREFUND\b)(?!.*\bREVERSAL\b)")

mask = df.description.str.strip().map(lambda s: bool(REFUND_ONLY.match(s)))
print("refund-only rows:", mask.sum())
print("rows mentioning REVERSAL:", df.description.str.contains(r"\bREVERSAL\b").sum())
print("overlap:", (mask & df.description.str.contains(r"\bREVERSAL\b")).sum())
print(df.loc[mask, "description"].head(3).to_list())
```

### Solution 8

This is the classic "password rules" idiom applied to reference codes. Each lookahead re-scans from
the same start position, so the three conditions are checked independently; the final
`[A-Z0-9]{6,10}` is what actually consumes the text.

```python
MIXED_CODE = re.compile(r"""
    \b
    (?=[A-Z0-9]{6,10}\b)      # right length, and only A-Z0-9 to the next boundary
    (?=[A-Z0-9]*[A-Z])        # at least one letter
    (?=[A-Z0-9]*\d)           # at least one digit
    [A-Z0-9]{6,10}\b
""", re.X)

hits = df.description.map(MIXED_CODE.findall)
print("codes found:", hits.map(len).sum(), "across", (hits.map(len) > 0).sum(), "rows")
print(sorted({c for row in hits for c in row})[:12])

for s in ["QAEDQECX", "232260", "7RT4EURT", "AB12"]:
    print(f"{s:>10} -> {bool(MIXED_CODE.fullmatch(s))}")
```

### Solution 9

`(?=%)` keeps the `%` out of the captured text, so `float()` works directly on the group. Compare
with `(\d+(?:\.\d+)?)%` — same rows, but here the assertion form generalises to "followed by `%` or
the word `percent`" without changing what you capture.

```python
PCT_RE = re.compile(r"\d+(?:\.\d+)?(?=\s?%)")

blob = df.notes + " " + df.description
pcts = blob.map(lambda s: [float(x) for x in PCT_RE.findall(s)])
print("rows with any percentage:", (pcts.map(len) > 0).sum())
print("rows with a rate above 5%:", pcts.map(lambda xs: any(x > 5 for x in xs)).sum())
print("distinct values:", sorted({x for row in pcts for x in row}))

# the FX rate is correctly ignored:
print(PCT_RE.findall("INTL PURCHASE FX RATE 1.0503 FEE 3% CARD ****0265"))
```

## Section 3 — Greedy vs lazy, splitting, backreferences

### Solution 10

The pair `(?<!\*)\*(?!\*)` means "a star with no star on either side", which skips the `****` mask.
The right-hand bound is a lookahead listing every token that can legally follow a merchant name —
that is what makes the lazy `.+?` stop in the right place instead of at the first space.

```python
CITY = r"(?:SEATTLE|NEW\ YORK|AUSTIN|CHICAGO|SAN\ FRANCISCO|BOSTON|DENVER|MIAMI|PORTLAND|ATLANTA)"
LONE_STAR = r"(?<!\*)\*(?!\*)"

MERCHANT_RE = re.compile(rf"""
    {LONE_STAR}\s*
    (?P<merchant>.+?)                       # lazy: stop as early as the bound allows
    \s+(?=                                  # right-hand boundary
        {CITY}\s+[A-Z]{{2}}\b | CARD\b | \*{{4}} | IBAN\b | DISPUTE\b
      | ORIG\b | SUB\b | ID: | FX\b | CONF\b | SEQ\b | REF\b | PPD\b | DIR\b
    )
""", re.X)

star_rows = df[df.description.str.contains(LONE_STAR)]
merchants = star_rows.description.str.extract(MERCHANT_RE)
print("star rows:", len(star_rows), "| parsed:", merchants.merchant.notna().sum())
print(sorted(merchants.merchant.dropna().unique()))

# what greedy does instead:
sample = star_rows.description.iloc[0]
print("\nsample :", sample)
print("greedy :", re.search(LONE_STAR + r"(.+)", sample).group(1))
print("lazy   :", MERCHANT_RE.search(sample).group("merchant"))
```

### Solution 11

`re.split` takes a *pattern*, not a character, which is what makes `\s{2,}` possible. `str.strip()`
first, otherwise the leading whitespace on some rows produces an empty leading token.

```python
tokens = df.description.str.strip().map(lambda s: re.split(r"\s{2,}", s))
print("rows that split into >1 field:", (tokens.map(len) > 1).sum())

sample = df.description[df.description.str.contains(r"\s{2,}")].iloc[0]
print("\nraw          :", repr(sample))
print("split on \\s{2,}:", re.split(r"\s{2,}", sample.strip()))
print("split on \\s+   :", re.split(r"\s+", sample.strip())[:8], "...")

# keeping the delimiter: a capturing group in the pattern returns it in the result
print("\nwith delimiters:", re.split(r"(\s{2,})", sample.strip())[:5])
```

### Solution 12

`\b(\w+)\s+\1\b` is the canonical doubled-word regex. The backreference is what makes this
impossible with a plain "find any two words" pattern — the engine must remember what group 1
captured and match that literal text again.

```python
DOUBLED = re.compile(r"\b(\w+)\s+\1\b", re.I)

hits = df.notes.map(lambda s: m.group(1) if (m := DOUBLED.search(s)) else None)
print("rows with a doubled word:", hits.notna().sum())
print(hits.value_counts().to_string())
print("\nexamples:")
for n in df.notes[hits.notna()].head(3):
    print("  ", n)

# \b matters:
print("\nthe theme ->", bool(DOUBLED.search("the theme")),
      "| the the ->", bool(DOUBLED.search("the the")))
```

### Solution 13

The pattern captures day/month/year and the replacement reorders them. Because the replacement is a
template rather than code, this stays a single vectorised `str.replace` over the whole column.

```python
DOTTED = r"\b(\d{2})\.(\d{2})\.(\d{4})\b"
fixed = df.notes.str.replace(DOTTED, r"\g<3>-\g<2>-\g<1>", regex=True)

changed = fixed != df.notes
print("rows rewritten:", changed.sum())
for before, after in zip(df.notes[changed].head(3), fixed[changed].head(3)):
    print("  -", before)
    print("  +", after, "\n")

# \g<n> vs \n when a digit follows:
print(re.sub(r"(\d{2})\.(\d{2})\.(\d{4})", r"\g<3>-\g<2>-\g<1>0", "24.03.2024"))
```

## Section 4 — Substitution and PII redaction

### Solution 14

The double `@` is rejected for free: the character class before `@` does not include `@`, so no start
position can produce a match. Requiring `(?:\.[A-Za-z0-9-]+)+` in the domain rejects `user@northwind`.
A replacement *callable* is the tool whenever the output depends on the matched text.

```python
EMAIL_RE = re.compile(r"\b(?P<user>[A-Za-z0-9._%+-]+)@(?P<domain>[A-Za-z0-9-]+(?:\.[A-Za-z0-9-]+)+)\b")

def mask_email(m):
    user = m.group("user")
    return f"{user[0]}{'*' * 3}@{m.group('domain')}"

valid = df.merchant_contact.map(lambda s: bool(EMAIL_RE.search(s)))
print("valid addresses in merchant_contact:", valid.sum())

looks_like_email = df.merchant_contact.str.contains("@")
print("rejected (contains @ but invalid):", (looks_like_email & ~valid).sum())
print(df.merchant_contact[looks_like_email & ~valid].unique()[:6])

df["contact_masked"] = df.merchant_contact.map(lambda s: EMAIL_RE.sub(mask_email, s))
print("\n", df.loc[valid, ["merchant_contact", "contact_masked"]].head(5).to_string(index=False))
```

### Solution 15

`\b` at both ends is doing the real work: without it, `\b\d{2}-\d{7}\b` would match inside a longer
digit run. The two grouping shapes are mutually exclusive once anchored on word boundaries, so
redacting one cannot damage the other — worth asserting rather than assuming.

```python
SSN_RE = re.compile(r"\b\d{3}-\d{2}-\d{4}\b")
EIN_RE = re.compile(r"\b\d{2}-\d{7}\b")

has_ssn = df.notes.map(lambda s: bool(SSN_RE.search(s)))
has_ein = df.notes.map(lambda s: bool(EIN_RE.search(s)))
print("rows with SSN:", has_ssn.sum(), "| rows with EIN:", has_ein.sum())

df["notes_safe"] = df.notes.map(lambda s: SSN_RE.sub("***-**-****", s))

print("\nbefore:", df.notes[has_ssn].iloc[0])
print("after :", df.notes_safe[has_ssn].iloc[0])
print("\nEIN row untouched:", df.notes_safe[has_ein].iloc[0])

assert df.notes_safe.map(lambda s: bool(SSN_RE.search(s))).sum() == 0
assert df.notes_safe.map(lambda s: bool(EIN_RE.search(s))).sum() == has_ein.sum()
print("\nredaction verified: 0 SSNs remain, all EINs preserved")
```

### Solution 16

Making every separator `[-.\s]?` is what collapses seven conventions into one pattern. The UK numbers
are handled by *checking the `intl` group* rather than by a second pattern — matching first and
branching afterwards keeps the regex small.

The stricter NANP rule is `[2-9]\d{2}` for the area code; this synthetic data uses random digits, so
the solution stays with `\d{3}` and flags the difference instead.

```python
US_PHONE = re.compile(r"""
    (?P<intl>\+\d{1,3})?[-.\s]?
    \(?(?P<area>\d{3})\)?[-.\s]?
    (?P<pre>\d{3})[-.\s]?
    (?P<line>\d{4})
    (?:\s*(?:x|ext\.?)\s*(?P<ext>\d{1,5}))?
""", re.X | re.I)

def normalize_phone(s):
    m = US_PHONE.search(s)
    if not m:
        return None
    g = m.groupdict()
    if g["intl"] and g["intl"] != "+1":
        return m.group(0).strip()                      # non-NANP: leave as-is
    out = f"+1-{g['area']}-{g['pre']}-{g['line']}"
    return out + (f" x{g['ext']}" if g["ext"] else "")

phones = df.merchant_contact[~df.merchant_contact.str.contains("@")]
norm_phones = phones.map(normalize_phone)
print("phone rows:", len(phones), "| normalised:", norm_phones.notna().sum())
print(pd.DataFrame({"raw": phones, "e164": norm_phones}).head(12).to_string(index=False))

strict = re.compile(r"\(?[2-9]\d{2}\)?[-.\s]?\d{3}[-.\s]?\d{4}")
print("\nwould pass a strict NANP area-code rule:",
      phones.map(lambda s: bool(strict.search(s))).sum(), "of", len(phones))
```

## Section 5 — The pandas `.str` regex API

### Solution 17

One vectorised call, two typed columns, `NaN` where a row has no match — this is almost always
preferable to `.apply(lambda s: re.search(...))` both in speed and in how the result lands in the
frame.

```python
extracted = df.description.str.extract(
    r"(?:(?<=\*\*\*\*)|(?<=CARD\ ))(?P<card>\d{4})", flags=re.X)
extracted["ref"] = df.description.str.extract(r"(?:REF#|REF\s|CASE#|CONF\s)(?P<ref>[A-Z0-9]+)")

out = pd.concat([df[["txn_id"]], extracted], axis=1)
print(out.notna().sum().to_string())
print()
print(out.dropna().head(6).to_string(index=False))
```

### Solution 18

`extractall` returns a MultiIndex of `(original row, match number)`, which is exactly what you need
to `unstack` or pivot. Rows with no match simply do not appear — that is the main behavioural
difference from `extract`.

```python
LABELLED = r"(?P<label>REF#|TERM:|SEQ\s|CASE#|ID:|CONF\s)(?P<code>[A-Z0-9]+)"
allcodes = df.description.str.extractall(LABELLED)
allcodes["label"] = allcodes["label"].str.strip().str.rstrip("#:")

print("total matches:", len(allcodes), "across", allcodes.index.get_level_values(0).nunique(), "rows")
print("\nrows with 2 matches:", (allcodes.groupby(level=0).size() == 2).sum())
print(allcodes.head(8).to_string())

wide = allcodes.reset_index(level=1, drop=True).pivot(columns="label", values="code")
print("\n", wide.head(6).to_string())
```

### Solution 19

A dict of `category -> pattern` plus `np.select` is the maintainable shape for rule-based labelling:
adding a merchant is a one-line change, and each rule stays independently testable. Order matters —
`np.select` takes the first matching condition — so put the specific rules before the generic ones.

```python
import numpy as np

RULES = {
    "coffee":        r"STARBUCKS|BLUE\s+BOTTLE",
    "grocery":       r"WHOLEFDS|WHOLE\s+FOODS",
    "transport":     r"\bUBER\b|\bLYFT\b",
    "fuel":          r"\bSHELL\b|\bEXXON\b",
    "travel":        r"DELTA\s+AIR|\bUNITED\s+AIR",
    "dining":        r"FRENCH\s+LAUNDRY|\bTST\*",
    "entertainment": r"NETFLIX|SPOTIFY|STEAMGAMES",
    "software":      r"APPLE\.COM|GOOGLE\s+\*CLOUD|ADOBE",
    "utilities":     r"COMCAST|CON\s+EDISON",
    "health":        r"CVS/PHARMACY",
    "home":          r"HOME\s+DEPOT",
    "retail":        r"AMZN|TARGET",
    # fallback tier: memo types that carry no merchant at all.
    # note \s+ rather than a single space - some rows arrived with mangled whitespace.
    "cash_transfer": r"CHECK\s+#|ATM\s+WITHDRAWAL|ONLINE\s+TRANSFER|DEPOSIT\s+PAYROLL",
}

conds = [df.description.str.contains(p, case=False, regex=True) for p in RULES.values()]
df["category"] = np.select(conds, list(RULES.keys()), default="uncategorised")

print(df.category.value_counts().to_string())
print("\nuncategorised share: {:.1%}".format((df.category == "uncategorised").mean()))

# a single space instead of \s+ silently loses the mangled rows:
naive = df.description.str.contains(r"ATM WITHDRAWAL|DEPOSIT PAYROLL|ONLINE TRANSFER", case=False)
robust = df.description.str.contains(r"ATM\s+WITHDRAWAL|DEPOSIT\s+PAYROLL|ONLINE\s+TRANSFER", case=False)
print(f"\nsingle-space rules: {naive.sum()} rows | \\s+ rules: {robust.sum()} rows")
```

## Section 6 — Capstone

### Solution 20

Two details make coverage 100% rather than ~85%: writing the literal as `POS\s+PURCHASE` (not a
single space) so the mangled rows still match, and bounding the lazy `merchant` with the city
alternation instead of hoping `.+?` stops in the right place.

In verbose mode a literal space must be escaped (`\ `) or written as `[ ]` — otherwise it is ignored.

```python
POS_RE = re.compile(rf"""
    ^\s*POS\s+PURCHASE\s+
    (?P<merchant>.+?)\s+
    (?P<city>{CITY})\s+
    (?P<state>[A-Z]{{2}})\s+
    CARD\s+(?P<card>\d{{4}})\s+
    REF\#(?P<ref>[A-Z0-9]+)\s+
    TERM:(?P<term>[A-Z0-9]+)\s*$
""", re.X)

pos_rows = df[df.description.str.contains(r"POS\s+PURCHASE")]
parsed_pos = pos_rows.description.str.extract(POS_RE)
print(f"POS PURCHASE rows: {len(pos_rows)} | parsed: {parsed_pos.merchant.notna().sum()}"
      f" ({parsed_pos.merchant.notna().mean():.0%})")
print(parsed_pos.dropna().head(6).to_string(index=False))
```

### Capstone, part 2 — the clean table

Everything above, assembled into one analytics-ready frame.

```python
clean = pd.DataFrame({
    "txn_id":     df.txn_id.map(normalize_id),
    "txn_id_raw": df.txn_id,
    "date":       df.date_iso,
    "amount":     df.amount,
    "ccy":        df.ccy,
    "category":   df.category,
    "card_last4": df.description.map(lambda s: m.group(0) if (m := CARD_RE.search(s)) else None),
    "ref":        df.description.str.extract(r"(?:REF#|REF\s|CASE#|CONF\s)([A-Z0-9]+)")[0],
    "contact":    df.contact_masked,
    "notes":      df.notes_safe,
})

print(clean.head(10).to_string(index=False))
print("\n--- data quality report ---")
report = pd.Series({
    "rows":                    len(clean),
    "ids normalised":          clean.txn_id.notna().sum(),
    "ids rejected":            clean.txn_id.isna().sum(),
    "dates parsed":            clean.date.notna().sum(),
    "amounts parsed":          clean.amount.notna().sum(),
    "currency recovered":      clean.ccy.notna().sum(),
    "card suffix found":       clean.card_last4.notna().sum(),
    "reference code found":    clean.ref.notna().sum(),
    "categorised":             (clean.category != "uncategorised").sum(),
    "SSNs remaining (must be 0)": clean.notes.map(lambda s: bool(SSN_RE.search(s))).sum(),
})
print(report.to_string())

clean.to_csv("finance_transactions_clean.csv", index=False)
print("\nwrote finance_transactions_clean.csv")
```

## Cheat sheet — what this notebook drilled

| tool | syntax | used in |
|---|---|---|
| named group | `(?P<name>...)` | Ex 2, 3, 17, 20 |
| non-capturing group | `(?:...)` | everywhere |
| alternation with anchors | `^(?:A|B|C)$` | Ex 2, 3 |
| lazy quantifier | `+?` `*?` | Ex 10, 20 |
| positive lookahead | `(?=...)` | Ex 7, 8, 9 |
| negative lookahead | `(?!...)` | Ex 7, 10 |
| positive lookbehind (fixed width) | `(?<=...)` | Ex 5 |
| negative lookbehind | `(?<!...)` | Ex 10 |
| backreference in a pattern | `\1` | Ex 12 |
| backreference in a replacement | `\g<1>` / `\1` | Ex 13 |
| replacement callable | `re.sub(p, func, s)` | Ex 14, 15 |
| verbose mode | `re.VERBOSE` / `re.X` | Ex 3, 4, 16, 20 |
| split on a pattern | `re.split` | Ex 11 |
| first match per row | `Series.str.extract` | Ex 17 |
| all matches per row | `Series.str.extractall` | Ex 18 |
| boolean mask | `Series.str.contains` | Ex 19 |

**Three habits worth keeping**

1. Anchor (`^...$` / `fullmatch`) when *validating*, don't anchor when *finding*.
2. Prefer a lazy quantifier plus an explicit right-hand boundary (a lookahead for the next known token) over a greedy `.*`.
3. Use `re.VERBOSE` the moment a pattern outgrows one line — a commented regex is a maintainable regex.
