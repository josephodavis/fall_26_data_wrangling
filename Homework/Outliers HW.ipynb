```python
import numpy as np
```

```python
waiting_times = [14, 15, 13, 14, 16, 13, 15, 14, 14, 15, 13, 14, 16, 14, 100]
waiting_times.sort()
print(waiting_times)
```

```python
mean = np.mean(waiting_times)
std = np.std(waiting_times, ddof=1)

z = (100 - mean) / std

ceiling = (15-1)/(15**0.5)

print(mean)
print(std)
print(z)
print(ceiling)

print(z / ceiling)
```

```python
Q1 = waiting_times[3]
Q3 = waiting_times[10]
IQR = Q3 - Q1

print(Q1, Q3)
print(IQR)
```

```python
fence1 = Q1-(1.5*IQR)
fence2 = Q3+(1.5*IQR)
print(fence1, fence2)
```

```python
median = np.median(waiting_times)

MAD = np.median(np.abs(waiting_times - median))

M = (0.6745*(100 - median)) / MAD
print(median)
print(MAD)
print(M)
```
