```python
list_triple = ["foo bar baz", "bar foo baz", "baz foo bar",
               "foo baz bar", "baz bar foo"]
```

```python
print(re.findall(r"^foo.*", "\n".join(list_triple)))
```

```python
print(re.findall(r"^[^foo].*", "\n".join(list_triple)))
```

```python
print(re.findall(r"^foo.*.bar$", "\n".join(list_triple)))
```

```python
ha_list = ["ha", "haha", "hahaha"]
```

```python
re.findall(r"\b(?:ha){1,2}\b", "\n".join(ha_list))
```
